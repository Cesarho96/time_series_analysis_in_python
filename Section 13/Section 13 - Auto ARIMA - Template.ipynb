### Packages

```python
import numpy as np
import pandas as pd
import scipy
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from statsmodels.tsa.arima.model import ARIMA
from arch import arch_model
import seaborn as sns
import yfinance
import warnings
warnings.filterwarnings("ignore")
sns.set()
```

### Loading the data

```python
raw_data = yfinance.download (tickers = "^GSPC ^FTSE ^N225 ^GDAXI", start = "1994-01-07", end = "2018-01-29", 
                              interval = "1d", group_by = 'ticker', auto_adjust = True, threads = True)
```

```python
df_comp = raw_data.copy()
```

```python
df_comp['spx'] = df_comp['^GSPC'].Close[:]
df_comp['dax'] = df_comp['^GDAXI'].Close[:]
df_comp['ftse'] = df_comp['^FTSE'].Close[:]
df_comp['nikkei'] = df_comp['^N225'].Close[:]
```

```python
df_comp = df_comp.iloc[1:]
del df_comp['^N225']
del df_comp['^GSPC']
del df_comp['^GDAXI']
del df_comp['^FTSE']
df_comp=df_comp.asfreq('b')
df_comp=df_comp.fillna(method='ffill')
```

### Creating Returns

```python
df_comp['ret_spx'] = df_comp.spx.pct_change(1)*100
df_comp['ret_ftse'] = df_comp.ftse.pct_change(1)*100
df_comp['ret_dax'] = df_comp.dax.pct_change(1)*100
df_comp['ret_nikkei'] = df_comp.nikkei.pct_change(1)*100
```

### Splitting the Data

```python
size = int(len(df_comp)*0.8)
df, df_test = df_comp.iloc[:size], df_comp.iloc[size:]
```

### Fitting a Model

```python
from pmdarima.arima import auto_arima
```

```python
model_auto = auto_arima(df.ret_ftse[1:])
```

```python
print(model_auto)
```

```python
model_auto.get_params()
```

```python
model_auto.summary()
```

SARIMAX is the most general model, it includes seasonality, auto regression, integration, moving average and exogenous factors, so auto_arima takes a sarimax model as the backbone.
Not seeing the seasonal components it suggest that this is a model without seasonal components, this is backed up by the model arguments we examined in the last video since they stated a seasonal order of (0,0,0,0). Thus almost our model could be ARIMAX than SARIMAX.  
Since the order of integration (d) is zero, this is a non-integrated model.  
Additionally, we didn't set an exogenous variable before fitting. We also didn't see one estimated among the coefficients, so it's safe to assume this is not a MAX model.  
Therefore, the most complex this model can be is an ordinary ARMA(4,5)  
The ARMA(4,5) wasn't our preferred model of choice when we conducted our initial manual analysis, we disregarded it because some of its coefficients were not significant, as we can clearly see in the summary table as well.  
A few quick comments:  
1.) The rules of model selection are rather "rules of thumb" than "fixed"  
2.) Auto ARIMA only considers a single feature - the AIC, the better the fit, regardless of the significance of some coefficients.  
3.) We could have easily overfitted while going through the models in our previous sections.  
4.) The default aruments of the method restrict the number of AR and MA components we are willing to include


### Important Arguments

[pmdarima.arima.auto_arima](https://alkaline-ml.com/pmdarima/modules/generated/pmdarima.arima.auto_arima.html#pmdarima.arima.auto_arima)

```python
model_auto = auto_arima(df_comp.ret_ftse[1:], exogenous = df_comp[['ret_spx', 'ret_dax', 'ret_nikkei']][1:], m = 5,
                       max_order = None, max_p = 7, max_q = 7, max_d = 2, max_P = 4, max_Q = 4, max_D = 2,
                       maxiter = 50, alpha = 0.05, n_jobs = -1, trend = 'ct', information_criterion = 'oob',
                       out_of_sample = int(len(df_comp)*0.2))

# Explanation of parameters:
# exogenous -> outside factors (e.g other time series)
# m -> seasonal cycle length
# max_order -> maximum amount of variables to be used in the regression (p + q)
# max_p -> maximum AR components
# max_q -> maximum MA components
# max_d -> maximum Integrations
# maxiter -> maximum iterations we're giving the model to converge the coefficients (becomes harder as the order increases)
# return_valid_fits -> whether or not the method should validate the results 
# alpha -> level of significance, default is 5%, which we should be using most of the time
# n_jobs -> how many models to fit at a time (-1 indicates "as many as possible")
# trend -> "ct" usually
# information_criterion -> 'aic', 'aicc', 'bic', 'hqic', 'oob' 
#        (Akaike Information Criterion, Corrected Akaike Information Criterion,
#        Bayesian Information Criterion, Hannan-Quinn Information Criterion, or
#        "out of bag"--for validation scoring--respectively)
# out_of_smaple_size -> validates the model selection (pass the entire dataset, and set 20% to be the out_of_sample_size)
```
